Los métodos kernel son un truco matemático para resolver problemas complejos organizando los datos de otra manera.

La idea básica
Imagina que tienes puntos rojos y azules mezclados en una hoja de papel (2D) y no puedes dibujarle una línea recta para separarlos.

El truco del kernel hace lo siguiente:

Levanta los puntos: Piensa que "eleva" la hoja a 3 dimensiones.

Corta con un plano: Al darles altura, ahora sí puedes pasar un papel plano entre los puntos rojos y azules para separarlos fácilmente.

Lo genial del kernel es que logra este cambio de dimensión con una fórmula matemática directa, sin gastar tiempo ni memoria calculando la posición de cada punto en 3D.
